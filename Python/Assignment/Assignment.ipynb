```python
num = int(input("Enter a number: "))

factorial = 1
if num < 0:
    print("Sorry, factorial does not exist for negative numbers")   
elif num == 0:
    print("The factorial of 0 is 1")    
else:
    for i in range(1,num + 1):
        factorial = factorial*i
    print("The factorial of",num,"is",factorial)    
    
```

16. Write a Python program to count the number of characters (character frequency) in a string

```python
text = input("Enter a word or sentence: ")
count = {}

for letter in text:          # go through each letter one by one
    if letter in count:      # if the letter is already counted
        count[letter] += 1   # add +1 to its count
    else:                    # if the letter is seen first time
        count[letter] = 1    # set count = 1

print("Letter Count:")
for letter, times in count.items():
    print(letter, "=", times)
```

22. Write a Python function to reverses a string if its length is a multiple of 4.

```python
def reverse_if_multiple_of_4(s):
    if len(s) % 4 == 0:
        return s[::-1]
    else:
        return s
    
print(reverse_if_multiple_of_4("abcd"))   
print(reverse_if_multiple_of_4("hello"))
```

58.Write a Python program to combine values in python list of dictionaries. Sample data: [{'item': 'item1', 'amount': 400}, {'item': 'item2', 'amount': 300}, o {'item': 'item1', 'amount': 750}]

```python
data = [
    {'item': 'item1', 'amount': 400},
    {'item': 'item2', 'amount': 300},
    {'item': 'item1', 'amount': 750}
]

result = {}

for d in data:
    item = d['item']
    amount = d['amount']
    
    if item in result:
        result[item] += amount   
    else:
        result[item] = amount    
print(result)
```

61. Write a Python function to calculate the factorial of a number (a nonnegative integer)

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))  # 120
```

63. Write a Python function to check whether a number is perfect or not.

```python
def is_perfect(num):
    divisors_sum = sum(i for i in range(1, num) if num % i == 0)
    return divisors_sum == num

print(is_perfect(6))   # True
print(is_perfect(28))  # True
print(is_perfect(10))  # False
```

89. How Do You Handle Exceptions with Try/Except/Finally in Python? Explain with coding snippets.
