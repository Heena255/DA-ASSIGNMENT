```python
import pandas as pd
sales = pd.read_csv(r"C:\Users\DELL\Desktop\TOPS\Numpy\Assessment\historical_automobile_sales.csv")
```

1. Q 1: Develop a Line chart using the functionality of pandas to show how
automobile sales fluctuate from year to year. 

```python
import matplotlib.pyplot as plt
print(sales.head())
sales.plot(x='Year', y='Automobile_Sales', kind='line',marker = 'o' , figsize= (10,6), title='Automobile Sales Over Years')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.show()

```

2. Plot different lines for categories of vehicle type and analyze the trend
to answer the question Is there a noticeable difference in sales trends
between different vehicle types during recession periods? 

```python
recession_sales = sales[sales['Recession'] == 1]
sales_by_type = recession_sales.groupby(['Year', 'Vehicle_Type'])['Automobile_Sales'].sum().unstack()
print(sales_by_type.head())
sales_by_type.plot(kind='line', marker='o', figsize=(12,6), title="Automobile Sales by Vehicle Type During Recession")
plt.xlabel("Year")
plt.ylabel("Number of Automobiles Sold")
plt.grid(True)
plt.legend(title="Vehicle Type")
plt.show()
```

3. Use the functionality of Seaborn Library to create a visualization to compare the sales trend per vehicle type for a recession period with a non- recession period. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sales_summary = sales.groupby(['Year', 'Vehicle_Type', 'Recession'])['Automobile_Sales'].sum().reset_index()
print(sales_summary.head())
plt.figure(figsize=(14,7))
sns.lineplot(
    data=sales_summary,
    x='Year',
    y='Automobile_Sales',
    hue='Vehicle_Type',       # Different lines for vehicle types
    style='Recession',        # Solid vs dashed lines for recession/non-recession
    markers=True,
    dashes={0: "", 1: (2,2)},  # Optional: dashed lines for recession
    marker='o' )
plt.title("Automobile Sales Trends by Vehicle Type: Recession vs Non-Recession")
plt.xlabel("Year")
plt.ylabel("Number of Automobiles Sold")
plt.grid(True)
plt.legend(title="Vehicle Type / Recession")
plt.show()
```

4. Now you want to compare the sales of different vehicle types during a recession and a non-recession period 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
avg_sales = sales.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
print(avg_sales)
plt.figure(figsize=(12,6))

sns.barplot(
    x='Vehicle_Type',
    y='Automobile_Sales',
    hue='Recession',   # Separate bars for recession (1) vs non-recession (0)
    data=avg_sales
)

plt.title("Average Automobile Sales by Vehicle Type: Recession vs Non-Recession")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Number of Automobiles Sold")
plt.grid(axis='y')
plt.legend(title="Recession")
plt.show()
```
