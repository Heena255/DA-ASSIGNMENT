Healthcare Industry:- 
Design a Python class ClinicAppointment that manages patient appointments in a clinic.
The system should have the following features:
➔ Book Appointment:
● Prompt for patient name, age, mobile number, and preferred doctor.
● Show time slots (10am, 11am, 12pm, 2pm, 3pm).
● Check slot availability and confirm booking.
➔ View/Cancel Appointment:
● Allow patient to view or cancel their appointment using mobile number.
➔ Doctor Availability:
● Maintain a maximum of 3 appointments per time slot per doctor.
➔ Data Persistence:
● Store appointments in memory only (no files/dbs required).

```python
# Simple Clinic Appointment System 

appointments = []   # all appointments in one list
time_slots = ["10am", "11am", "12pm", "2pm", "3pm"]

# function to count how many patients are booked for a doctor at a slot
def count_slot(doctor, slot):
    count = 0
    for a in appointments:
        if a["doctor"] == doctor and a["slot"] == slot:
            count += 1
    return count

# book appointment
def book_appointment():
    name = input("Enter patient name: ")
    age = input("Enter age: ")
    mobile = input("Enter mobile number: ")
    doctor = input("Enter doctor name: ")
    print("Available slots:", time_slots)
    slot = input("Enter slot: ")

    if slot not in time_slots:
        print("❌ Wrong slot")
        return

    # check if patient already booked
    for a in appointments:
        if a["mobile"] == mobile:
            print("❌ You already booked an appointment.")
            return

    # check if slot full (max 3)
    if count_slot(doctor, slot) >= 3:
        print("❌ This slot is full for Dr.", doctor)
        return

    # save appointment
    appointments.append({"name": name, "age": age, "mobile": mobile,
                         "doctor": doctor, "slot": slot})
    print("✅ Appointment booked with Dr.", doctor, "at", slot)

# view appointment
def view_appointment():
    mobile = input("Enter mobile number: ")
    for a in appointments:
        if a["mobile"] == mobile:
            print("📅 Appointment:", a["name"], "Age:", a["age"],
                  "Doctor:", a["doctor"], "Slot:", a["slot"])
            return
    print("❌ No appointment found.")

# cancel appointment
def cancel_appointment():
    mobile = input("Enter mobile number: ")
    for a in appointments:
        if a["mobile"] == mobile:
            appointments.remove(a)
            print("✅ Appointment cancelled.")
            return
    print("❌ No appointment found.")

# show availability
def show_availability():
    doctor = input("Enter doctor name: ")
    for slot in time_slots:
        left = 3 - count_slot(doctor, slot)
        print(slot, "->", left, "slots left")

# ------------------- Menu -------------------
while True:
    print("\n--- Clinic Menu ---")
    print("1. Book Appointment")
    print("2. View Appointment")
    print("3. Cancel Appointment")
    print("4. Doctor Availability")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        book_appointment()
    elif choice == "2":
        view_appointment()
    elif choice == "3":
        cancel_appointment()
    elif choice == "4":
        show_availability()
    elif choice == "5":
        print("Goodbye!")
        break
    else:
        print("❌ Wrong choice, try again.")
```

School Management System:- 
Design a Python class SchoolManagement that helps manage student admissions and records. The system should support:
➔ New Admission:
● Collect student name, age, class (1–12), and guardian's mobile number.
● Assign a unique student ID automatically.
● Validate age: must be between 5 and 18.
● Validate mobile number: must be 10 digits.
➔ View Student Details:
● Allow lookup using student ID.
➔ Update Student Info:
● Update mobile number or class.
➔ Remove Student Record:
● Remove a student using their student ID.
➔ Exit System

```python
class SchoolManagement:
    def __init__(self):
        self.students = {}   # {student_id: details}
        self.next_id = 1

    def run(self):
        while True:
            print("\n--- School Management Menu ---")
            print("1. New Admission")
            print("2. View Student Details")
            print("3. Update Student Info")
            print("4. Remove Student Record")
            print("5. Exit")

            choice = input("Enter choice: ")

            # ---------------- New Admission ----------------
            if choice == "1":
                name = input("Enter student name: ")
                age = int(input("Enter age: "))
                student_class = int(input("Enter class (1-12): "))
                mobile = input("Enter guardian mobile (10 digits): ")

                if age < 5 or age > 18:
                    print("❌ Age must be between 5 and 18.")
                elif student_class < 1 or student_class > 12:
                    print("❌ Class must be between 1 and 12.")
                elif len(mobile) != 10 or not mobile.isdigit():
                    print("❌ Mobile must be 10 digits.")
                else:
                    sid = self.next_id
                    self.next_id += 1
                    self.students[sid] = {
                        "name": name,
                        "age": age,
                        "class": student_class,
                        "mobile": mobile
                    }
                    print("✅ Admission successful. Student ID =", sid)

            # ---------------- View Student ----------------
            elif choice == "2":
                sid = int(input("Enter Student ID: "))
                if sid in self.students:
                    s = self.students[sid]
                    print("📘 Student ID:", sid)
                    print("Name:", s["name"])
                    print("Age:", s["age"])
                    print("Class:", s["class"])
                    print("Guardian Mobile:", s["mobile"])
                else:
                    print("❌ Student not found.")

            # ---------------- Update Student ----------------
            elif choice == "3":
                sid = int(input("Enter Student ID: "))
                if sid in self.students:
                    print("What do you want to update?")
                    print("1. Mobile Number")
                    print("2. Class")
                    option = input("Enter option: ")

                    if option == "1":
                        new_mobile = input("Enter new mobile number: ")
                        if len(new_mobile) == 10 and new_mobile.isdigit():
                            self.students[sid]["mobile"] = new_mobile
                            print("✅ Mobile updated.")
                        else:
                            print("❌ Invalid mobile number.")
                    elif option == "2":
                        new_class = int(input("Enter new class (1-12): "))
                        if 1 <= new_class <= 12:
                            self.students[sid]["class"] = new_class
                            print("✅ Class updated.")
                        else:
                            print("❌ Invalid class.")
                    else:
                        print("❌ Wrong option.")
                else:
                    print("❌ Student not found.")

            # ---------------- Remove Student ----------------
            elif choice == "4":
                sid = int(input("Enter Student ID: "))
                if sid in self.students:
                    del self.students[sid]
                    print("✅ Student removed.")
                else:
                    print("❌ Student not found.")

            # ---------------- Exit ----------------
            elif choice == "5":
                print("Goodbye! Exiting system.")
                break

            else:
                print("❌ Invalid choice. Try again.")


# Run the program
school = SchoolManagement()
school.run()
```

Transport Reservation System (Bus Ticketing)
Design a Python class BusReservation that simulates a basic bus ticket booking system. Features should include:
➔ Show Available Routes:
● Predefined city routes with fixed prices.
● Example: "Mumbai to Pune - ₹500", "Delhi to Jaipur - ₹600", etc.
➔ Book Ticket:
● Enter passenger name, age, mobile, and route.
● Assign seat number (max 40 per bus per route).
● Generate a unique ticket ID.
➔ View Ticket:
● Lookup using ticket ID.
➔ Cancel Ticket:
●Exit.

```python
class BusReservation:
    def __init__(self):
        # predefined routes with prices
        self.routes = {
            "Mumbai to Pune": 500,
            "Delhi to Jaipur": 600,
            "Bangalore to Mysore": 400,
            "Chennai to Coimbatore": 450
        }

        self.tickets = {}   # {ticket_id: details}
        self.next_ticket_id = 1
        self.seats = {}     # {route: [list of booked seats]}

        # initialize seats for each route
        for r in self.routes:
            self.seats[r] = []

    def run(self):
        while True:
            print("\n--- Bus Reservation Menu ---")
            print("1. Show Available Routes")
            print("2. Book Ticket")
            print("3. View Ticket")
            print("4. Cancel Ticket")
            print("5. Exit")

            choice = input("Enter choice: ")

            # ---------------- Show Routes ----------------
            if choice == "1":
                print("\nAvailable Routes:")
                for route, price in self.routes.items():
                    print(f"{route} - ₹{price}")

            # ---------------- Book Ticket ----------------
            elif choice == "2":
                name = input("Enter passenger name: ")
                age = int(input("Enter age: "))
                mobile = input("Enter mobile number: ")

                print("\nChoose Route:")
                for i, route in enumerate(self.routes.keys(), start=1):
                    print(i, route, "-", "₹", self.routes[route])

                route_choice = int(input("Enter route number: "))
                route_list = list(self.routes.keys())

                if route_choice < 1 or route_choice > len(route_list):
                    print("❌ Invalid route.")
                    continue

                route = route_list[route_choice - 1]

                # check seat availability
                if len(self.seats[route]) >= 40:
                    print("❌ No seats left for this route.")
                    continue

                seat_no = len(self.seats[route]) + 1
                self.seats[route].append(seat_no)

                ticket_id = self.next_ticket_id
                self.next_ticket_id += 1

                self.tickets[ticket_id] = {
                    "name": name,
                    "age": age,
                    "mobile": mobile,
                    "route": route,
                    "seat_no": seat_no,
                    "price": self.routes[route]
                }

                print(f"✅ Ticket Booked! Ticket ID = {ticket_id}, Seat No = {seat_no}")

            # ---------------- View Ticket ----------------
            elif choice == "3":
                tid = int(input("Enter Ticket ID: "))
                if tid in self.tickets:
                    t = self.tickets[tid]
                    print("\n🎫 Ticket Details")
                    print("Ticket ID:", tid)
                    print("Name:", t["name"])
                    print("Age:", t["age"])
                    print("Mobile:", t["mobile"])
                    print("Route:", t["route"])
                    print("Seat No:", t["seat_no"])
                    print("Price: ₹", t["price"])
                else:
                    print("❌ Ticket not found.")

            # ---------------- Cancel Ticket ----------------
            elif choice == "4":
                tid = int(input("Enter Ticket ID: "))
                if tid in self.tickets:
                    route = self.tickets[tid]["route"]
                    seat = self.tickets[tid]["seat_no"]
                    self.seats[route].remove(seat)
                    del self.tickets[tid]
                    print("✅ Ticket cancelled successfully.")
                else:
                    print("❌ Ticket not found.")

            # ---------------- Exit ----------------
            elif choice == "5":
                print("Goodbye! Exiting system.")
                break

            else:
                print("❌ Invalid choice. Try again.")


# Run the system
bus = BusReservation()
bus.run()
```
